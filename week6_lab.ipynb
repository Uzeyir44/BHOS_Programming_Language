Question 1: Grade Checker



Create a Tkinter window titled “Grade Checker” with:



A Label that says “Enter your grade (0-100):” with font ("Arial", 14, "bold"), foreground white, background navy.

An Entry widget where the user types a grade.

A Button labeled “Check”.

A result Label that is empty at the start.

When the button is clicked, the command function should:



Read the grade from the Entry.

If grade ≥ 50, show “Passed” with green foreground.

If grade < 50, show “Failed” with red foreground.

If the input is not a number, show “Invalid input” with orange foreground.

```python
import tkinter as tk

def check():
    g = int(grade.get())
    if g >= 50:
        result.config(text="Passed", fg="green")
    else:
        result.config(text="Failed", fg="red")

root = tk.Tk()
root.title("Grade Checker")

l1 = tk.Label(text="Enter your grade (0-100):", font=("Arial", 14, "bold"), fg="white", bg="navy")
grade = tk.Entry(root)
check_button = tk.Button(text="Check", command=check)
result = tk.Label(text="")

l1.pack(pady="20")
grade.pack(pady="20")
check_button.pack(pady="20")
result.pack(pady="20")

root.mainloop()
```

Question 2: Class Average Calculator



Create a window with:



Three Entry widgets for Math, Physics, and Programming grades, each with a Label beside it (font ("Courier", 12), background lightgray).

A Button labeled “Calculate Average” with background blue and foreground white.

A result Label with font ("Helvetica", 16, "italic").

When the button is clicked, the command function should:



Calculate the average of the three grades.

Display “Average: X.XX” in the result label.

Change the result label’s background by average: green if ≥ 90, yellow if 70-89, red if below 70.

Change the button text to “Recalculate” after the first click using config().

```python
import tkinter as tk

def average():
    av = (int(m.get()) + int(ph.get()) + int(pl.get()))/3
    
    if av >= 90:
        result.config(bg="green")
    elif av >= 70:
        result.config(bg="yellow")
    else:
        result.config(bg="red")
    
    average_button.config(text="Recalculate")

root = tk.Tk()

math = tk.Label(text="Math", font=("Courier", 12), bg="lightgray")
physics = tk.Label(text="Physics", font=("Courier", 12), bg="lightgray")
programming = tk.Label(text="Progrmming", font=("Courier", 12), bg="lightgray")

average_button = tk.Button(text="Calculate Average", bg="blue", fg="white", command=average)

result = tk.Label(text="Result", font=("Helvetica", 16, "italic"))

m = tk.Entry(root)
ph = tk.Entry(root)
pl = tk.Entry(root)

math.pack(pady="5")
m.pack(pady="5")
physics.pack(pady="5")
ph.pack(pady="5")
programming.pack(pady="5")
pl.pack(pady="5")

average_button.pack(pady="10")
result.pack(pady="5")

root.mainloop()
```
