Write a function calculate_area(length, width) that returns the area of a rectangle; call it and print the result

```python
def calculate_area(length, width):
    return (length*width)

print(calculate_area(100, 50))
```

Add a default value for width=1 so the function can also calculate area for a square with just length

```python
def calculate_area(length, width=1):
    return (length*width)

print(calculate_area(100))
```

Write a function total_sum(*args) that returns the sum of any number of numbers passed to it

```python
def total_sum(*args):
    return (sum(args))

print(total_sum(1, 2, 3, 4, 5))
```

Write a function print_profile(**kwargs) that prints each key-value pair passed to it (e.g., name, age, city)

```python
def print_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} -> {value}")

print_profile(name="Katy", age=19, major="Electrical Engineering", future="Bright")
```

Demonstrate scope: create a global variable, then write a function that tries to modify it — once without global and once with global, printing the result each time

```python
var = 10

def local():
    var += 1
    return  var

def glob():
    global var
    var += 1
    return  var

try:
    print(local())
except:
    print("Works with local varibles, so returns an error")

print(glob())
```

Use a lambda function inside sorted() to sort a list of tuples [("Bob", 23), ("Alice", 20)] by age

```python
ls = [("Bob", 23), ("Alice", 20)]

ls = sorted(ls, key=lambda x: x[1])

print(ls)
```
