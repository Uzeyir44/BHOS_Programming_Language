# Create a DataFrame using the following dictionary

```python
students = {
    "Name": ["Ali", "Aysel", "Murad", "Leyla"],
    "Math": [80, 90, 75, 85],
    "Python": [85, 95, 80, 78],
    "Database": [70, 88, 82, 78]
}

students = pd.DataFrame(students)
print(students)
```

# Add two new columns to the DataFrame: Total containing the sum of each students' scores, and average

```python
temp = students.iloc[0:len(students), 1:len(students.columns)].copy()
total = []
average = []


for i in range(len(temp)):
    total.append(temp.loc[i].sum())
    average.append(round(temp.loc[i].mean(), 1))
    
students["Total"] = pd.Series(total)
students["Average"] = pd.Series(average)

print(students)
```
