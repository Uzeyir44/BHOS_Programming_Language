# Initialization of elements

```python
import numpy as np

id = np.arange(101, 106)
names = ("Aysel", "Murad", "Leyla", "Tural", "Nigar")
scores = [78, 92, 65, 88, 71] 
```

# Display the third student's name, ID, and score.

```python
print(f"{id[2]}     {names[2]}      {scores[2]}")
```

# Display every student as: ID - Name: Score.

```python
for i in range(id.shape[0]):
    print(f"{id[i]}     {names[i]}      {scores[i]}")
```

# Change Murad's score to 95 and display the list.

```python
scores[names.index("Murad")] = 95
print(scores)
```

# Add a new score of 84 to the list.

```python
scores.append(84)
```

# Add the ID 106 to the array.

```python
id = np.append(id, 106)
```

# Try to change the first name in the tuple, then create a new tuple that includes "Farid".

```python
names[0] = "Farid"
```

```python
new_names = names + ("Farid",)
```

# Find the highest, lowest, and average score (2 decimals).

```python
print(f"Highest score: {max(scores)}")
print(f"Lowest score: {min(scores)}")
print(f"Average score: {sum(scores)/len(scores):.2f}")
```

# Display scores from highest to lowest without changing the original list.

```python
sorted_scores = sorted(scores)[::-1]

for i in sorted_scores:
    print(i)
```

# Create a list of passing scores (70 or above) and count them.

```python
passed_students = [x for x in scores if x >= 70]

print(passed_students)
print(len(passed_students))
```

# Combine ID, name, and score for all six students, then find the one with the highest score.

```python
combined = []

for i in range(id.shape[0]):
    temp = []
    temp.append(int(id[i]))
    temp.append(new_names[i])
    temp.append(scores[i])
    
    combined.append(temp)
    
maxi = -1
index = 0

for i in range(len(combined)):
    if combined[i][2] > maxi:
        maxi = combined[i][2]
        index = i

print(f"Student with the highest score: {combined[index]}")
```
