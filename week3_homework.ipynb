# Grading System

```python
score = int(input("Enter your score: "))

if score>=91 and score<=100:
    print("A")
elif score>=81:
    print("B")
elif score>=71:
    print("C")
elif score>=61:
    print("D")
elif score>=51:
    print("E")
else:
    print("F")
```

# Multiplication Table

```python
n = int(input("Enter the number: "))

for i in range(1, 11):
    print(f"{n} x {i} = {n*i}")
```

# Password

```python
password = "password"
i = 1

while 1:
    user = input("Enter the password: ")
    
    if user == password:
        print("Correct Password")
        break
    elif i >= 3:
            print("Exceeded the limit. Relaunch the program to try again")
            break
    else:
        print("Wrong Password. Try one more time")
        
    
        
    i += 1
```
