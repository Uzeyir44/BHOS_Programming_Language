Add a class attribute total_books that increments every time a new Book is created; add a @classmethod get_total_books() to return it

```python
class Book():
    total_books = 0
    
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price
        Book.total_books += 1
    
    def __str__(self):
        return (f"{self.title} by {self.author} ${self.price}")
    
    @classmethod
    def get_total_books(cls):
        return (cls.total_books)
```

Add a @staticmethod apply_discount(price, percent) that returns the discounted price

```python
class Book():
    total_books = 0
    
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self._price = price
        Book.total_books += 1
    
    def __str__(self):
        return (f"{self.title} by {self.author} ${self.price}")
    
    @classmethod
    def get_total_books(cls):
        return (cls.total_books)
    
    @property
    def price(self):
        return (self._price)
    
    @price.setter
    def price(self, value):
        if value < 0:
            raise ValueError
        
        self._price = value
    
    @staticmethod
    def apply_discount(price, percent):
        return (price*(100-percent)/100)
```
