Create a dictionary representing a student (name, age, grades as a list); print each
field using .get()

```python
std = {
    "name": "Ziya",
    "age": 18,
    "grades": []
}

print(std.get("name"), std.get("age"), std.get("grades"))
```

Add a new key "email"
, update the age, then remove "grades" using pop()

```python
std["email"] = "zorucov@gmail.com"
std["age"] = 19
std.pop("grades")
```

Loop through the dictionary using .items() and print each pair formatted as key ->
value

```python
for key, value in std.items():
    print(f"{key}->{value}")
```

```python
a = set([1, 2, 3,  4, 5])
b = set([3, 4, 5, 6, 7])

print(a.union(b), a|b)
print(a.intersection(b), a&b)
print(a.difference(b), a-b)
```

Create a pandas Series of five exam scores with student names as the index; print
the score of one student by label and the highest score

```python
import pandas as pd

scores = [99, 99, 98, 44, 55]
idx = ["Ziya", "Semed", "Leyla", "Royala", "Qanira"]

srs = pd.Series(scores, index=idx)
print(srs)
```

Create a DataFrame from a dictionary with columns name, age, and grade for at
least four students

```python
students = {
    "name": ["Ziya", "Semed", "Leyla", "Qanira"],
    "age": [18, 18, 25, 18],
    "grade": [100, 100, 100, 100]
}

frame = pd.DataFrame(students)
print(frame)
```

Select the name column, then use loc and iloc to select the first two rows

```python
print(frame.loc[0:1, "name"])
```

Filter the DataFrame to show only students older than 20, then add a new column passed using a condition on grade

```python
old = frame[frame["age"] > 20]
frame["passed"] = frame["grade"] > 75
print(old)
print(frame)
```

Drop the age column and print describe() for the remaining numeric columns

```python
new = frame.drop(columns=["age"])
print(new.describe())
```
