#Task 1

```python
score = int(input("Enter your score: "))

if score>=91 and score<=100:
    print("A")
elif score>=81:
    print("B")
elif score>=71:
    print("C")
elif score>=61:
    print("D")
elif score>=51:
    print("E")
else:
    print("F")
```

#Task 2

```python
n = int(input("Enter number: "))

for i in range(1, n+1):
    if i%3==0 and i%5==0:
        print(3, 5)
    elif i%3==0:
        print(3)
    elif i%5==0:
        print(5)
    else:
        print(i)
        
print("The sum of the even numbers is:", sum([x for x in range(1, n+1) if x%2==0]))
print("The number of elements divisible by 3: ", len([x for x in range(1, n+1) if x%3==0]))
```
