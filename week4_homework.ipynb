Create a list of 5 favorite movies; append() one, remove() one, then sort() and
print it

```python
ls = ["movie1", "movie2", "movie3", "movie4", "movie5"]
ls.append("Spider-Man")
ls.remove("movie2")
ls.sort()
print(ls)
```

Using indexing/slicing, print the first movie, last movie, and first three movies

```python
print(ls[0])
print(ls[-1])
print(ls[:3])
```

Create a list of numbers 1–20; use a list comprehension to get only the even
numbers

```python
nums = [i for i in range(1, 21) if i%2==0]
print(nums)
```

Create a 3×3 array using np.arange(1, 10).reshape(3, 3); print the second row, the
last column, and the element at row 1, column 2

```python
m = np.arange(1, 10).reshape(3, 3)
print(m[1])
print(m[:, 2])
print(m[0, 1])
```

Create an array of numbers 1–10; use vectorized operations to multiply every
element by 2, then use a boolean mask to print only the elements greater than 10

```python
new = np.arange(1, 11)
new = new*2
print(new)
print(new[new > 10])
```
